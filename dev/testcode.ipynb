```python
def augment(df):
    data = []
    alphabet = list("abcdefghijklmnopqrstuvwxyz")
    seed = rnd.randint(0,4)
    for sample, label in df.values:
        data.append((sample,label))
        chars = list(sample)
        pos = rnd.randint(0,len(chars)-1)
        if seed ==0: #random deletion
            chars.remove(chars[pos])
        elif seed ==1: #random insertion
            chars.insert(pos, rnd.choice(alphabet))
        elif seed==2: #random swap
            temp = chars[pos]
            if pos==len(chars)-1:
                chars[pos]=chars[pos-1]
                chars[pos-1]=temp
            else:
                chars[pos]=chars[pos+1]
                chars[pos+1]=temp
        elif seed==3: #scramble
            rnd.shuffle(chars)
        elif seed==4: #random replacement
            char = rnd.choice(alphabet)
            chars[pos]=char
        data.append(("".join(chars),label))
    return pd.DataFrame(data)

import pandas as pd

df = pd.read_csv('datasets/confirmations.csv')
new_df = pd.DataFrame(augment(df))
data = new_df[0].values
labels = new_df[1].values
print(data)
print(labels)
print(df.columns.values)

```

```python
import statistics

artist_genre={}
for artist in individual_artists:
    modal_genre = statistics.mode(df[df['artists'].apply(lambda x: artist in x)]['track_genre'].values)
    artist_genre[artist.lower()]=modal_genre.lower()
```

```python
from nltk.corpus import wordnet
import random as rnd, string
def synonym_set(text):
    augments = [text]
    rnd.seed(42)
    words = text.split()
    #replace with 3 synonyms
    for _ in range(3):
        for word in words:
            synonyms = set()
            for syn in wordnet.synsets(word):
                for lemma in syn.lemmas():
                    #ensures replacement
                    if lemma.name()!=word:
                        synonyms.add(lemma.name())
            synonyms=list(synonyms)
            #replace the word
            if synonyms:
                augments.append(text.replace(word,rnd.choice(synonyms)))
    return augments
chars = list('word')
aug_chars=chars
aug_chars.insert(rnd.randint(0,len(chars)),rnd.choice(string.ascii_letters))
aug_word = "".join(aug_chars)
print(aug_word)
```
