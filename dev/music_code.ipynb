```python
import pandas as pd
import statistics
import joblib
```

```python
df = pd.read_csv('datasets/music_genre.csv')
df=df.dropna()
df['artists'] = df['artists'].apply(lambda x: x.split(';'))
joblib.dump(df,'music/df.joblib')
df['track_genre']
```

```python

individual_artists = set()
for a in df['artists'].values:
    for art in a:
        individual_artists.add(art)

artist_genre={}
for artist in individual_artists:
    modal_genre = statistics.mode(df[df['artists'].apply(lambda x: artist in x)]['track_genre'].values)
    artist_genre[artist.lower()]=modal_genre.lower()

genre_df = pd.DataFrame(artist_genre.items(), columns=['artist','genre'])
joblib.dump(genre_df,'music/genre_df.joblib')
```

```python
from Api import *
from Artist import *
api = Api()
artist = 'David Bowie'
artist,shows = api.request_artist(artist)

if shows>0:
    output = ''
    print(f"{artist} has {shows} coming up. Would you like to see these events?")
    user_input = input()
    if user_input=='yes':
        events = api.request_event(artist)
        output=''
        for _ in events:
            output+=f'''They are playing on {events[_]['date']} in {events[_]['location']}.\n'''
    else: output = 'ok'
else:
    output=''
    print(f" I'm sorry, {artist} has no upcoming shows.\nSearching for similar artists...")
    a = Artist(artist)
    new_artist = a.rand_similar()
    while not (api.request_artist(new_artist)[1]>0):
        new_artist = a.rand_similar()
    print(f"{new_artist} has {shows} coming up. Would you like to see these events?")
    user_input = input()
    if user_input=='yes':
        events = api.request_event(new_artist)
        output=''
        for idx in events:
            output+=f'''They are playing on {events[idx]['date']} in {events[idx]['location']}.\n'''
    else: output = 'ok'
print(f"{output if output else f'''Sorry, I can't find any upcoming {a.genre} shows.'''}")
```

```python
import re
text=['i want to see some rock music','heavy metal']
m=r'''(?:.*?\b(i|we)(?:'d|'ll|\b)\s*(?:will|would(?:\slike)?|want|like)?\s*(?:to\s*)?(?:go\s*(?:to\s*)?)?(?:see(?:\s*some)?)?\s*)?
(?P<music>.*?(?=on|$|in|music))?'''
for t in text:
    match = re.search(m,t.lower(), re.IGNORECASE | re.VERBOSE)
    if match:
        print(match.group('music'))
```
